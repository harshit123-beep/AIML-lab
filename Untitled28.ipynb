```python
# Twitter Sentiment Analysis using SVM


# Import Libraries
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Create Dataset
tweets = {
    "Tweet": [
        "I love this mobile",
        "Worst experience ever",
        "Amazing customer support",
        "Very bad quality",
        "I am happy with the product",
        "This app is terrible",
        "Fantastic performance",
        "I hate this service",
        "Best purchase",
        "Not worth the money"
    ],

    "Sentiment": [
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative"
    ]
}

# Convert into DataFrame
data = pd.DataFrame(tweets)

# Input and Output
X = data["Tweet"]
y = data["Sentiment"]

# Convert Text to Numbers
vectorizer = CountVectorizer()
X_vector = vectorizer.fit_transform(X)

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_vector,
    y,
    test_size=0.2,
    random_state=1
)

# Create SVM Model
model = SVC(kernel='linear')

# Train Model
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Test Custom Tweet
sample = ["I really like this application"]

sample_vector = vectorizer.transform(sample)

result = model.predict(sample_vector)

print("Prediction:", result[0])
```

```python

# Customer Segmentation using K-Means


# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Create Dataset
customer_data = {
    "Age": [18, 22, 25, 30, 35, 40, 45, 50],
    "Spending": [20, 25, 30, 50, 55, 70, 80, 90]
}

# Convert into DataFrame
data = pd.DataFrame(customer_data)

# Select Features
X = data[["Age", "Spending"]]

# Create KMeans Model
kmeans = KMeans(n_clusters=3, random_state=0)

# Train Model
kmeans.fit(X)

# Assign Clusters
data["Cluster"] = kmeans.labels_

# Print Output
print(data)

# Plot Graph
plt.figure(figsize=(7,5))

plt.scatter(
    data["Age"],
    data["Spending"],
    c=data["Cluster"],
    s=100
)

# Plot Centroids
plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    marker='X',
    s=250
)

# Labels
plt.xlabel("Age")
plt.ylabel("Spending Score")
plt.title("K-Means Customer Clustering")

# Show Graph
plt.show()
```
